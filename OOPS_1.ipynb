```python
class Home: #Home is the class name and Class is the keyword
    a = 100;
    def Bedroom(obj):
        print("Bedroom")

    def Lobby(obj):
        print("Lobby")

ram = Home() #Home() is a constructor and ram is user defined object

ram.Bedroom()
ram.Lobby()
print(ram.a)
```

```python
class Home: #Home is the class name and Class is the keyword
    a = 100;
    def Bedroom(self): #the first argument in member function is compulsory, usually self
        print("Bedroom")

    def Lobby(self):
        print("Lobby")

ram = Home() #Home() is a constructor and ram is user defined object

ram.Bedroom()
ram.Lobby()
print(ram.a)
```

```python
#why we need to have 'self' in the member function
class Home: #Home is the class name and Class is the keyword
    a = 100;
    def Bedroom(self): #the first argument in member function is compulsory, usually self
        print("Bedroom")

    def Lobby(self):
        self.Bedroom() #an function cannot be called without an object, that self is the object
        print("Lobby" , self.a)

        #to access even a member function or a member variable -> we need to access it using an object

ram = Home() #Home() is a constructor and ram is user defined object

ram.Bedroom() 
ram.Lobby()
print(ram.a)
```

```python
#why we need to have 'self' in the member function
class Home: #Home is the class name and Class is the keyword
    a = 100;
    def Bedroom(self): #the first argument in member function is compulsory, usually self, self is not a keyword
        #self is an object_name but not a strict keyword
        print("Bedroom")

    def Lobby(self,a):
        self.Bedroom() #an function cannot be called without an object, that self is the object
        print("Lobby" , self.a)
        print(f'Lobby {a}') # see how a and self.a are different

        #to access even a member function or a member variable -> we need to access it using an object

ram = Home() #Home() is a constructor and ram is user defined object

ram.Bedroom() 
ram.Lobby(12)
print(ram.a)
```

```python
#how to create static member function in python
#python does not have default static keyword
#a static variable has only one instace and no multiple copies are not allowed
class chd:
    def information(self):
        print("chd")
obj = chd()
obj.information()
```
