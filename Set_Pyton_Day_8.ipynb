```python
s = {1,12,2,1,2,3}
print(s)
#gives only unique values in any order
```

```python
#traversing a set
#to traverse a non-linear data type we use foreach loop
for item in s:
    print(item)
```

```python
s1= {12,42,3,213}
s2={43,35,1435,3,432}
print(s1.intersection(s2))
```

```python
print(s1.intersection(s2))
```

```python
print(s1)
```

```python
s1.remove(12) 
s1.discard(12)
#remove gives error if element does not exist in set whiel discard gives no error
```

```python
set_one={1,32,3}
set_two={1,3}
print(set_two.issubset(set_one))
```

```python
print(set_one.issuperset(set_two))
```

```python
#merging two sets together (Update Method)

set1={1,2,43,22,22,42,'hello'}
set2={2,3,2,44}

set1.update(set2)
print(set1)
```

```python
#Difference method 
set1={1,2,43,22,22,42,'hello'}
set2={2,3,2,44}

s = set1.difference(set2)
print(s)
```

```python
#Difference method (another syntax) new_set = set_one - set_two;
set1={1,2,43,22,22,42,'hello'}
set2={2,3,2,44}

s = set1-set2
print(s)
```

```python
#Difference_update method (another syntax) new_set = set_one - set_two;
'''
while the differce method creates a new varialble to store the set
the difference_update does not create a new variable, it just updates the initial variable
'''
set1={1,2,43,22,22,42,'hello'}
set2={2,3,2,44}

# set1.difference_update(set2)
set1 -= set2
print(set1)
```

```python
#another example
set1={1,2,43,22,22,42,'hello'}
set2={2,3,2,44}
set3={'hello'}
set4= set1.difference(set2,set3)
print(set4)
```
