```python
#Multilevel Inheritence
'''Basic Example '''
class A:
    def TestA(self):
        print(f"I am from class A")
class B(A):
    def TestB(self):
        print("From B")
              
class C(B):
    def TestC(self):
        print(f'From C')
obj = C()
obj.TestA()
obj.TestB()
```

```python
#another example
class A:
    r = None
    m = None
    def first_val(self):
        self.r = int(input("Enter first val"))

class B(A):
    def sec_val(self):
        self.m = int(input("Enter second val"))

class C(B):
    def result(self):
        self.first_val()
        self.sec_val()
        print(f'{self.r+self.m}')

ojb = C()
ojb.result()
    
```

```python
#Multiple Inheritence
class A:
    a = None
    def ist(self):
        self.a = int(input("enter ist"))

class B:
    b=None
    def iind(self):
        self.b = int(input("Enter iind"))
class C(A,B):
    def result(self):
        self.ist()
        self.iind()
        print(self.a+self.b)
obj = C()
obj.result()
```

```python
#Method order Resolution -> read on your own
class A
    a = None
    def fun(self):
        print("A")

class B:
    b=None
    def fun(self):
        print("B")
class C(A,B):
    def result(self):
        self.fun()
obj = C()
obj.fun()
```

```python
class A:
    def fna(self):
        print("A")

class B(A):
    def fnb(self):
        print("B")

class C(A):
    def fnc(self):
        print("C")

class D(B):
    def fnd(self):
        print("D")

obj1 = B()
obj2 = D()
obj1.fna()
obj1.fnb()
obj2.fna()
obj2.fnb()
```

```python
#Nested Class
'''
Class inside a class
'''
class Human:
    def Info(self):
        print("HUMANS ARE FOUND ON EARTH")
    class Male:
        def Info(self):
            print("male")
    class Female:
        def Info(self):
            print("female")
human = Human()
male = Human.Male()
male.Info()
human.Info()
```

```python
#Nested Class
'''
Class inside a class
'''
class Human:
    a = "test"
    
    def Info(self):
        print("HUMANS ARE FOUND ON EARTH", self.a)
    class Male:
        def Info(self):
            print("male")
        
    class Female:
        def Info(self):
            print("female")
human = Human()
male = Human.Male()
female = Human.Female()
female.Info()
male.Info()
human.Info()
```

```python
#Nested Class
'''
Class inside a class
'''
class Human:
    _p = "p"
    a = "test"
    
    def Info(self):
        print("HUMANS ARE FOUND ON EARTH", self.a)
    class Male:
        __p ="private"
        def Info(self):
            print("male", Human.a)
        
    class Female:
        def Info(self):
            print("female", Human._p)
human = Human()
male = Human.Male()
female = Human.Female()
female.Info()
male.Info()
human.Info()
human.Female().Info()
#protected variable cannot be nested
```

```python
class T:
    p =100
    def display(self):
        print("P value is", T.p)
    class Gt:
        def hm(self,outer):
            print("p value is", outer.p)
obj=T()
obj1 = T.Gt()
obj.display()
obj1.hm(obj)
```

```python
class T:
    __p =100
    def display(self):
        print("P value is", T.__p)
    class Gt:
        def hm(self,outer):
            print("p value is", outer._T__p) # see this name mangling
obj=T()
obj1 = T.Gt()
obj.display()
obj1.hm(obj)
```
