                                                        PARAMETERISED FUNCTION

```python
#Example 1;
def sum(a : int, b: int, c: int): #a,b,c are parameters
    return a+b+c
    
print(f"Sum is {sum(1,3,4)}")
#1,3,4 are acutal values passed according to the number of parameters
```

```python
#Example 2;
def sum(a : int, b: int, c: int): #a,b,c are parameters
    return a+b+c
    
print(f"Sum is {sum(3,4)}")
#passing lesser or more arguments gives error
```

```python
#Example 3; ASSIGING DEFAULT VALUES
def sum(a : int, b: int , c: int=7): #a,b,c are parameters
    return a+b+c
    
print(f"Sum is {sum(3,4)}")
#passing lesser or more arguments gives error
```

```python
#Example 4; ASSIGING DEFAULT VALUES
def sum(a : int = 3, b: int = 4, c: int=7): #a,b,c are parameters
    return a+b+c
    
print(f"Sum is {sum()}")
#passing lesser or more arguments gives error
```

```python
#Example 4; INPUT AT RUNTIME:
def sum(a : int = 3, b: int = 4, c: int=7): #a,b,c are parameters
    x,y,z = map(int,input("Enter the values of a,b,c").split())
    return x+y+z
print(f"Sum is {sum()}")
#passing lesser or more arguments gives error
```

```python
# the * method
def sum(*a):
    print(type(a))
    for item in a:
        print(f"the value of {item}")
sum(1,2,3,4,5)
```

```python
# THE DOUBLE STAR ** (multiple parameter + key-value pair)
def hr(**p):
    print(type(p))
    print(p)
    for key, value in p.items(): 
        #see how the p.items() prints key and values in a single loop
        print(f"The key is {key} : and value is {value}")
hr(name='harh', rollno=12, size =113)
```

```python
#Unpacking Dictionary Using Parameter double astrisk
def ef(Name,Dept,Id):
    print("Employee Name", Name)
    print("Employee Dept", Dept)
    print("Employee Id", Id)
data = {"Name": 'one', "Dept": 2, "Id":'threeeezz'}
ef(**data) #**data unpacks and assign the variables as values of the keys
```

```python
#Unpacking Dictionary Using Parameter double astrisk
def ef(Name,Dept,Id):
    print("Employee Name", Name)
    print("Employee Dept", Dept)
    print("Employee Id", Id)
data = {Name: 'one', "DEpt": 2, "Id":'threeeezz'}
#the key names should be in " " and must be exactly same to the parameters declared
ef(**data) #**data unpacks and assign the variables as values of the keys
```
