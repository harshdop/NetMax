Dictionary - {key and value pair data type}
1. Also know as JSON (JavaScript Object Notation)
       {
       "name": "parmas",
       "roll": 23
       
       }
2. Declared using {}
3. No duplicacies of keys
4. No indexing
5. We can have list, set, tuple in dictionaries
6. It is ordered

```python
dic = {"name" : "harsh",
       "age" : 12
      }
print(dic)
```

```python
#Values are mutable
dic["name"] = 23
print(dic)
```

```python
#although keys are immutable but we can declare them as
dic = {"name" : "harsh",
       "age" : 12,
       "name" : "deep"
      }
print(dic)

#name will be overritten with the last key-value pair 
```

```python
inp : int = int(input("enter any number"))
dic = { inp : "harsh",
       "age" : 12,
       "name" : "deep",
        '' : "empty",
        " " : "null"
      }
print(dic)
print(dic[inp])
print('age')
print(dic[''])
print(dic[" "])
```

```python
#extracting keys and values
s_list = dic.keys()
print(s_list)
s_list=list(s_list)

#printing the key
for i in s_list:
    print(i)
print(type(s_list))

#printing the value
print("printing the values")
for i in range (len(s_list)):
    print(dic[s_list[i]])

```

```python
dic = { "name" : "harsh",
       "age" : 12,
       "name" : "deep",
        '1' : "empty",
        " 2" : "null"
      }
from prettytable import PrettyTable
obj = PrettyTable()

print(dic.keys())
obj.field_names = list(dic.keys())
# obj.add_row(1,2,3,4,5)  THE ADD_ROW ACCEPTS A LIST
obj.add_row([1,2,3,4,])
print(obj)
```

```python
# popping an element
w = {"Name" : "harsh",
     "Roll" : 23
    }
w.pop("Name");
print(w)
```

```python
#updating list 1 with list 2
w = {"Name" : "harsh",
     "Roll" : 23
    }
d = { "Name" : "zarsh",
       "age" : 12,
       "name" : "deep",
        '1' : "empty",
        " 2" : "null"
      }
w.update(d)

```

```python
# THE SET DEFAULT METHOD
w = {"Name" : "harsh",
     "Roll" : 23
    }
# adds a row with key value pair i
w.setdefault("Dept", "CDSSA")
print(w)

```

```python

w = {"Name" : "harsh",
     "Roll" : 23
    }
# the key-value is updated iff it does not exist initially
z = w.setdefault("Name", "Rohan")
print(w)
print(z)
```
