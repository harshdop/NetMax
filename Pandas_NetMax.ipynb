```python

df['OrderID']=df['OrderID'].fillna(df['OrderID'].mean())
df

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Age'].notnull().sum()

df['Product'] = df['Product'].fillna(df['Product'].mode()[0])
# mode will return a series of elements as multiple mode might exist, so it is necessary to provide an index such as [0]
df.head()
```

```python
df['CustomerName'] = df['CustomerName'].fillna(df['CustomerName'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Product'] = df['Product'].fillna(df['Product'].mode()[0])
df['Price'] = df['Price'].fillna(method = 'ffill')
df['Quantity'] = df['Quantity'].fillna(method = 'bfill')
df.isnull().sum()
```

```python
#co-relation
relation = df[['Age','Price','Quantity']].corr()
type(relation)
relation
```

```python
import seaborn as sns
sns.heatmap(relation, annot = True)
```
