OPERATOR : To perform an operation is called an Operator

```python
''' Types of Operators
1. Arthematic Operator (+ - * / %)
2. Assignment Operator (= += -= *= /= %=)
3. Membership Operator (in , not in)
4. Logical Operator (and , or , not)
5. Comparision/Relational Operator (> < >= <= != ==)
6. Identity Operator (is , is not)
'''
```

```python
a = 10;
a += 10;
print(a)
```

```python
#Membership operator (usually used with string, tuples and dictionaries)
s = "my name is harsh"
print("pankaj" in s)
```

```python
#Membership operator
s = "my name is harsh"
print("pankaj" not in s)
```

```python
#Membership operator

try:
    s = str(input("Enter your String"))
    if "pankaj" not in s :
        print("hello")
    else :
        print("syke")
except Exception as err:
    print(err)
```

```python
#Logical Operator (and, or, not)

try: 
    a = 100;
    print(a>110 or a<=120)
    
except Exception as err:
    print(err)
```

```python
#Logical Operator NOT
#NOT reverses the symbol i.e true to false and false to tre

try: 
    a = 100;
    print(not(a>110 or a<=120))
    #the condition is actually true (a<=120) but not reversed the true to false
    #hence the final output is False
    
except Exception as err:
    print(err)

finally:
    print("..............")
```

```python
#Comparision And Logical Operator

try: 
    a = 100;
    b=120
    print(a!=b)
    #the condition is actually true (a<=120) but not reversed the true to false
    #hence the final output is False
    
except Exception as err:
    print(err)

finally:
    print("..............")
```

```python
#Identity Operator

try: 
    a = [12];
    b = [12];
    #the thing to note here is that a is b, compare the refrences and not directly the values
    # the integers and small values are accessed via a quickly accessible storage space, 

    ''' 
    a,b refer to same memory when a,b are intergers and == 12;
    list a and list b are different memory locations, hence the false we get
    '''
    print( a is b) # works as ==
    
    
except Exception as err:
    print(err)

finally:
    print("..............")
```

```python
""" CONDITION STATEMENTS (DECISION STATEMENTS) """
'''
1. if statement
2 if-else statement
3. elif statement (else-if)
4. nested if-else

'''

try:
    a = input("Enter A");
    a = int(a);
    if (a>10):
        print("inside if statement")
    else :
        print("inside else statement")

    
except Exception as err:
    print(err)

finally:
    print("..............")
    
```

```python
""" CONDITION STATEMENTS (DECISION STATEMENTS) """
'''
1. if statement
2 if-else statement
3. elif statement (else-if)
4. nested if-else

'''

try:
    a,b= map(int,input("Enter the value of A and B").split())
    a = int(a);
    if (a == b):
        print("inside if statement")
    elif (a>b):
        pass #pass will skip out the if-else if it runs
    else :
        print("inside else statement")

    
except Exception as err:
    print(err)

finally:
    print("..............")
    
```

```python
#nested if-else (condition inside a condition)

try: 
    n : int = int(input("Enter the number"))
    #the user entered input is by default considered to be a stirng
    if (n>400 and n<500):
        if(n%2 ==0):
            print("Number is even")
        else:
            print("Number is odd")
    else:
        print("Enter number between 400 and 500 exclusive")

except Exception as err:
    print(err)

finally:
    print("..............")
    

```
