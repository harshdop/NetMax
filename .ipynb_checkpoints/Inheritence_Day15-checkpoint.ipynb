To create a new class from an existing class in called inheritence "https://www.geeksforgeeks.org/python/inheritance-in-python/"
    TYPES OF INHERITENCE
    a. Single Level inheritence -> one parent 
    b. Multilevel Inheritence -> a series of parent -> base -> subbase' -> subbase''
    c. Multiple Inheritence (READ HOW PYTHON AVOIDS DIAMOND PROBLEM OF AMBIGUITY) -> multiple parents have a single child
    d. Hierarichal Inheritence
    e. Hybrid Inheritence -> Mixture of two or more than two types
    
    
    

```python
#Single level inheritence
class A:
    def A(self):
        print("I am in class A")
class B(A):
    def B(self):
        self.A()
        print(f"I am in class B")

ojb = A()
ojb.B() # 'A' class has no member B()
```

```python
#Single level inheritence
""" BASE CAN ACCESS ALL OF PARENT, BUT PARENT CANNOT ACCESS MEMBERS OF BASE """
class A:
    def A(self):
        self.B()
        print("I am in class A")
        
class B(A):
    def B(self):
        
        print(f"I am in class B")

ojb = B()
ojb.A() # 'A' class has no member B()
```
