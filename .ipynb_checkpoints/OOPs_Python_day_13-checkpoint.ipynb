```python
#Static member function
class student:
    def information(self): 
        print(f"I am in student")

s1 = student()
s1.information()
```

```python
#Static member function
class student:
    def information(): #remove the self -> the function is static
        print(f"I am in student")

student.information()
```

```python
#access specifiers in python:
#the access spcification abities of python are not absolute
'''Private Members
__ double underscore -> private
_ single underscore -> protected
no underscore -> public'''
class dishes:
    def __private(self):
        print(f'I am in private')
    def public(self,a):
        print(f'I am in public {a}')

class cup:
    a = 100;
    def pr(self):
        self.public(a)
    
    

obj = dishes()
obj.public() #The private members cannot be accesed outside the class
c = cup()
c.pr()
```

```python
#access specifiers in python:
#the access spcification abities of python are not absolute
'''Private Members
__ double underscore -> private
_ single underscore -> protected
no underscore -> public'''
class dishes:
    def __private(self):
        print(f'I am in private')
    def public(self):
        self.__private()
        print("In public")


    

obj = dishes()
obj.public() #The private members cannot be accesed outside the class
```

```python
class stu:
    Roll = 12 # the default data member in python in public
    def info(self):
        print("Roll number is", self.Roll)
nstu = stu()
nstu.Roll = 100
nstu.info()
print(f'roll number is {nstu.Roll}')
print(f'roll number is {stu.Roll}')
```

```python
class stu:
    __Roll = 12 # the default data member in python in public
    def info(self):
        print("Roll number is", self.__Roll)
        
nstu = stu()
stu.__Roll = 100
nstu.info()
```

```python
#Encapsulation in python
'''
1. Hiding data + Controlling Access
2. You don't anyone to change data directly
3. To change data -> they have to use methods that you control

'''

class person:
    __age = 0
    def set_age(self,myage):
        if (myage > 0 and myage<=110):
            self.__age = myage
        else:
            print(f'Age cannot be negative and above 110')

    def get_age(self):
        print(f'Your age is {self.__age}')
        return self.__age
       

p1 = person()
p1.__age = 100;

print(p1.__dict__)
p1.get_age()

```

CONSTRUCTOR IN PYTHON
1. It is used to intialize your class data member, member function and object
2. Called automatically whenever an object is created
3. Constructors dont return any data member value
4. Constructor inside a class is declared with the help of __init__ but to call it we use class name
5. Constructor overloading is not possible in python (like we do in c++, Java, .Net, php)
6. ONLY ONE CONSTRUCTOR IS ALLOWED PER CLASS

Types of constructor
a.Default constructor (without any parameters)
b.Parameterized Constuctor (accepts parameters)
WE DONT HAVE COPY CONSTRUCTOR -> a copy constructor takes an oject as an input, the create an object we again need another constructor, but python does not allow constructor overloading

```python
a = [1,2,3]
b =a
b.append(2)
print(a)
```

```python
a = [1,2,3]
b =a[0:len(a)]
c = a[:]
b.append(2)
print(a)
print(b)
print(c)
```
