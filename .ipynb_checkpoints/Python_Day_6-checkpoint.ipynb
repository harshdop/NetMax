```python
'''
Loop : self-repetition of a task till a condition is true is known as a loop
Now this condition might be finite or infinite

Loops reduce LOC
Loops are used to program complex logics into code
Size of the code file gets reduced
Smaller file improved Execution speed -> increased efficiency

Loops were made for linear data structures (continuous memory allocation), such as array, lists, stacks , queues.
Loops help to develop logics for complex problems.

'''
```

```python
''' Types of loops in python:
1. for loop (runs on membership operator)
2. while loop (runs on boolean or conditions)
'''
```

```python
#for i in range (start, end, step)
for i in range(-5,-1,+1):
    print(i)
```

```python
a = int(input("enter a"))
fact = 1;
if(a<0):
    print(fact)
else:
    for i in  range (a, 0, -1):
        fact = fact*i;
    
print(fact)
        
```
