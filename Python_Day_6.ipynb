"BRANCHING STATEMENTS IN LOOPS" (optional statements)
1. Break -> stops the execution of program
2. continue -> skips the execution of a block of code if a condition is matched
3. goto -> jump to a specific statement (not available in python)
   

```python
### for loop using branching statement ->break

try:

    n = int(input("Enter loop size"))
    for i in range (n):
       
        if(i==5):
            break
        print(i)

except Exception as e:
    print(e)

finally:
    print(">>>>>>><<<<<<<<<")
```

```python
### for loop using branching statement ->break

try:

    n = int(input("Enter loop size"))
    for i in range (n):
       
        if(i==5):
            continue
            #see how print(5) will get skipped
        print("the value is",i)

except Exception as e:
    print(e)

finally:
    print(">>>>>>><<<<<<<<<")
```

```python
### for loop using branching statement ->break

try:

    n = int(input("Enter loop size"))
    for i in range (n,0,-1):
       
        
        print("the value is",i)

except Exception as e:
    print(e)

finally:
    print(">>>>>>><<<<<<<<<")
```
