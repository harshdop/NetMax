```python
"""                                             VERY VERY IMPORTANT CONCEPT 
    1. Non-local only and only works for nested functions
    2. Global is used in case when fucntion and varibale exist
"""
def tm():
    pizza = 200
    print(f"I am in function tm , i eat {pizza} pizza")
    def km():
         nonlocal pizza # the non-local does not access the local pizza
                        # non-local only works for nested functions
        
         pizza = 500;
         print(f"I am in function km i eat {pizza} pizza")
    km()
    print(f"tm also want {pizza} pizza")
tm()
    
        
    
```
