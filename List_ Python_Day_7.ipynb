```python
#length of the list
x : int = len(w)

for I in range (x):
    print(w[I], end=" ")

start : int = 0;

print("")


print("PRINTING USING WHILE LOOP")
while ( start<x):
    print(w[start], end= " " )
    start = start +1
    
```
