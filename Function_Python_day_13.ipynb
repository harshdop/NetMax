```python
'''
Function generator : instead of producing the output all at once, the fucntion generator allows you to prduce the output one by one.
    Instead of creating and storing values all at once in the memory, it created an object known as the gneretor object.
    The generator object stores the function execution state and produces value one by one.
        yield() is used instead of return
        and generator_obj = func() -> func output is given to the generator object 
    a.saves object
    b.handles large (infinite) object
    c.quick setup
    d.clean up
'''
```

```python
def afunc(x):
    for i in range (x):
        yield(i)

obj = afunc(7)
print(next(obj)) #produces output one by one
print(next(obj))
```

```python
''' The date and time function'''
import datetime
#which was the day on 10Oct2003
z = datetime.datetime(2003,6,5) #syntax(year, month, date)

print(z.strftime("%a")) #to print day in small consise

print(z.strftime("%A")) #to print day in full

print(z.strftime("%w")) #to print index of day (Sunday is 0)

print(z.strftime("%d")) #prints only date

print(z.strftime("%D")) #prints complete date

print(z.strftime("%y")) #prints year in short

print(z.strftime("%Y")) #prints complete year

print(z.strftime("%b"))#prints the month in short

print(z.strftime("%B"))#prints complete month

print(z.strftime("%m"))#prints the month index number (January is 1)


```
