A FUNCTION THAT RETURNS A VALUE IS CALLED A RETURN FUNCTION

```python
def fact(a : int):
    if(a == 0):
        return 1
    return a * (fact(a-1))
fact(4)
```

```python
def name():
    x = input("enter your name")
    return x;

print(f"your name is {name()}")
```

```python
def one(a,b):
    c = a+b;
    return c

def two(w):
    d = one(12,4)
    k=w-d;
    return k

ans = two(100)
print(f"Your result is {ans}")

#here function two is dependent on function one
```

                                                    FUNCTION DECORATOR

```python
#without using function decorator calling manually:
def fd(hi):
    def mn():
        hi()
    return mn

def gk():
    print("how are you")

xt = fd(gk)
xt()
```

```python
#using function decorator 

def fd(hi):
    def mn():
        hi()
    return mn
# @fd

def gk():
    print("how are you")

gk()



'''
Conditions of decorator:
1. decorator is not a default keyword;
2. the function after @ is the decorator function;
3. a decorator function must have atleast one parameter 
4. the parameter of the decorator must be a function itself
5. a decorator function must have a another function defined inside it;
6. inside the another function we must call the parameter function
7. always return the another function


'''
```

```python
def decorator(guard):
    def wrapper():
        print("Guard checking")
        guard()
        print("All is well")
    return wrapper

@decorator
def guard():
    print("checking")

guard()
```

```python
def decorator(guard):
    def wrapper():
        guard()
        print("Guard checking")
        # guard()?
        print("All is well")
    return wrapper

@decorator
def authenticate():
    print("checking")

authenticate()
```

```python
def decorator(math):
    def logic():
        pi_value = math()
        radius = int(input("Enter the radius"))
        # result = (pi_value*radius*radius)
        # how to write power
        result = pi_value*(radius**2)
        return result
    
    return logic

# @decorator
def x():
    return 3.14

y = decorator(x)
print(y())
```
