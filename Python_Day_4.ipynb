EXCEPTION HANDLING 

```python
print("hello")
print(xyx) #we will get an error as xyx is not defined
print("bye")
```

```python
#but we want xyx too.. we use error handling mechanism
print("hello")

try :
    print(xyx) #always keep your code in the try block
except Exception as err:
    print(err);
    
print("bye")

#see how try and catch avoid catastrophic failure due to error
```

```python
#but we want xyx too.. we use error handling mechanism
print("hello")

try :
    xyx = "niggazz"
    print(xyx) #always keep your code in the try block
except Exception as err:
    print(err);
    
print("bye")

#see how try and catch avoid catastrophic failure due to error
```

```python
try: 
    a,b = map(int,input("enter the value for a,b").split())
    print(a+b, "is the result")
    
except Exception as err:
    print(err)

print("Done")
              
```

```python
# using the FINALLY block
#finally always run.. always

try: 
    a,b = map(int,input("enter the value for a,b").split())
    print(a+b, "is the result")
    
except Exception as err:
    print(err)

finally:
    print("Finally is use to cleanup code execution")

    

print("Done")
              
```

```python
try:
    a,b = map(int,input("enter the value of a and b").split())
    c = int(a/b)
    print(c)
except Exception as err:
    print(err)
finally:
    print("........")
```
