```python
#19. Select 'customer_name' and 'purchase_amount' for customers with purchase_amount between 100 and 200 (inclusive).
df.loc[(df['purchase_amount']>=100) & (df['purchase_amount']<=200), ['customer_name','purchase_amount']]
'''
SEE how in pandas we donot and cannot use 'and' but we have to use '&'
    WHY WE GET THE ERROR: ValueError: The truth value of a Series is ambiguous. Use a.empty, a.bool(), a.item(), a.any() or a.all().
    df['purchase_amount'] >= 100 creates a series which each element True or False, so python tries to get a singlular boolean value for the 
    entire series, which does not make sense.
    & is actually a bit-wise operator, but pandas overwrites the & such that it compares row by row, and not once for the entire seires
    that is why in pandas filtering we use '&' insted of 'and'.
'''
```
